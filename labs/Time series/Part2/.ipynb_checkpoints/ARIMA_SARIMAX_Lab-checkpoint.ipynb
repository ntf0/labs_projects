## 0. Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.set_printoptions(precision=4)
sns.set(font_scale=1.5)
plt.style.use('fivethirtyeight')

%matplotlib inline
%config InlineBackend.figure_format = 'retina'
```

```python
from statsmodels.tsa.stattools import acf, pacf, adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tools.eval_measures import rmse
from pmdarima import auto_arima
```

<div class="alert alert-info"><strong>What is the below function?:</strong> Function for running the augmented Dickey-Fuller test (adfuller).</div>

```python
def adf_test(series, title=''):
    """
    Pass in a time series and an optional title, returns an ADF report
    """
    print(f'Augmented Dickey-Fuller Test: {title}')
    result = adfuller(series.dropna(),autolag='AIC') # .dropna() handles differenced data
    
    labels = ['ADF test statistic','p-value','# lags used','# observations']
    out = pd.Series(result[0:4],index=labels)

    for key,val in result[4].items():
        out[f'critical value ({key})']=val
        
    print(out.to_string())          # .to_string() removes the line "dtype: float64"
    
    if result[1] <= 0.05:
        print("Strong evidence against the null hypothesis")
        print("Reject the null hypothesis")
        print("Data has no unit root and is stationary")
    else:
        print("Weak evidence against the null hypothesis")
        print("Fail to reject the null hypothesis")
        print("Data has a unit root and is non-stationary")
```

<div class="alert alert-info"><strong>What is the below function?:</strong> plot ACF and PACF beside each other.</div>

```python
def autocorr_plots(y, lags=None):
    fig, ax = plt.subplots(ncols=2, figsize=(12, 4), sharey=True)
    plot_acf(y, lags=lags, ax=ax[0])
    plot_pacf(y, lags=lags, ax=ax[1])
    return fig, ax
```

# Part A - ARIMA

We'll start simple: treat this as a general time series and build a non-seasonal ARIMA model. Follow the same workflow: 

> **Stationarity -> Differencing -> ACF/PACF -> Fit -> Evaluate**.


### A1. Check stationarity

```python
# Run adf_test on the passenger series. Is it stationary?
adf_test(df['thousands_of_passengers'])
```

### A2. Difference until stationary

If the series isn't stationary, difference it (`.diff()`) and re-run the ADF test. Keep track of how many differences it took - that's your `d`.

```python
# Difference once and re-test
passenger_diff1 = df['thousands_of_passengers'].diff().dropna()

adf_test(passenger_diff1)
```

```python
# If needed, difference a second time and re-test
passenger_diff2 = df['thousands_of_passengers'].diff().diff().dropna()

adf_test(passenger_diff2)
```

**Write down your `d` value here:**

`d = 2`

### A3. ACF / PACF on the differenced series

Use `autocorr_plots()` on your *stationary* (differenced) series.

```python
# Plot ACF and PACF on the differenced series, lags=24
autocorr_plots(passenger_diff2, lags=24);
```

**Based on the plots above, what would you guess for `p` and `q`?**

- p (AR terms): 4
- q (MA terms): 2

### A4. Confirm with auto_arima

Now check your manual guess against `auto_arima`. Keep `seasonal=False` as we are not modeling seasonality in Part A.

```python
auto_model = auto_arima(
    df['thousands_of_passengers'],
    seasonal=False,
    trace=True
)

print(auto_model.summary())
```

```python
auto_model.order
```

### Observation and Comparison

The manual ARIMA analysis suggested **ARIMA(4, 2, 2)**:

* **p = 4:** estimated from the PACF plot.
* **d = 2:** the first difference was still slightly non-stationary according to the ADF test (**p-value = 0.0542 > 0.05**), so a second difference was required.
* **q = 2:** estimated from the ACF plot.

In comparison, `auto_arima` selected **ARIMA(4, 1, 3)** based on its automatic model-selection process and AIC.

The results are similar in identifying **p = 4**, but differ in the selected values of **d** and **q**. This is expected because the manual approach relies on interpreting the ADF, ACF, and PACF results, while `auto_arima` automatically searches and evaluates candidate models.

For the following modeling steps, **ARIMA(4, 1, 3)** will be used based on the `auto_arima` result, and its forecasting performance will be evaluated on the test set.


### A5. Train, Test, and Split

Hold out the last 12 months as your test set.

```python
# Split into df_train (all but last 12 rows) and df_test (last 12 rows)
df_train = df.iloc[:-12]
df_test = df.iloc[-12:]
```

### A6. Fit the ARIMA model

Use the (p,d,q) order from A4.

```python
# Fit ARIMA(p,d,q) on the training set
model = ARIMA(df_train['thousands_of_passengers'], order=(4, 1, 3))
model_fit = model.fit()
```

### A7. Predict on the test set

```python
# Predict over the test period
predictions = model_fit.forecast(steps=len(df_test))

predictions
```

```python
# Plot the test data vs predictions
plt.figure(figsize=(12, 5))

plt.plot(df_test.index, df_test['thousands_of_passengers'], label='Actual')

plt.plot(df_test.index, predictions, label='Predicted')

plt.xlabel('Month')
plt.ylabel('Thousands of Passengers')
plt.title('ARIMA: Actual vs Predicted')
plt.legend()
plt.show()
```

Compute RMSE for the ARIMA model. Also print the mean of the series so you have something to compare the error against.

# Part B - SARIMA

Now let's actually look at *why* the ARIMA forecast missed. This is where we introduce seasonality.


### B1. Seasonal decomposition

Run `seasonal_decompose` on the **original** (undifferenced) series and plot it.

```python
#YOUR CODE HERE
from statsmodels.tsa.seasonal import seasonal_decompose

decomposition = seasonal_decompose(
    df['thousands_of_passengers'],
    model='additive',
    period=12
)

decomposition.plot()
plt.show()
```

**What do you see in the seasonal component?** Is there a repeating pattern? What's the period (how many months before it repeats)?

**Observation**

> The seasonal component shows a clear and strong repeating pattern. Passenger numbers rise to a similar peak and then fall to a similar low each year.
Repeating pattern: Yes
Period: 12 months
This indicates strong yearly (annual) seasonality in the passenger data.

Therefore:
` m=12 `

### B2. Search for a seasonal order with auto_arima

Run `auto_arima` again, this time with `seasonal=True` and `m=12` (12 months per cycle).

```python
#YOUR CODE HERE
sarima_auto_model = auto_arima(
    df['thousands_of_passengers'],
    seasonal=True,
    m=12,
    trace=True
)

print(sarima_auto_model.summary())
```

**Write down the order and seasonal_order suggested:**

- order (p,d,q) = (2, 1, 1)
- seasonal_order (P,D,Q,m) = (0, 1, 0, 12) 

### B3. Fit the SARIMA model

Use `SARIMAX` with both `order` and `seasonal_order`, fit on `df_train`.

```python
#YOUR CODE HERE
sarima_model = SARIMAX(
    df_train['thousands_of_passengers'],
    order=sarima_auto_model.order,
    seasonal_order=sarima_auto_model.seasonal_order
)

sarima_fit = sarima_model.fit()
```

### B4. Predict on the test set, with confidence intervals

This time, also pull the confidence interval so you can plot forecast uncertainty.

```python
#YOUR CODE HERE
forecast = sarima_fit.get_forecast(steps=len(df_test))

sarima_predictions = forecast.predicted_mean
conf_int = forecast.conf_int()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df_test.index,
    df_test['thousands_of_passengers'],
    label='Actual'
)

plt.plot(
    df_test.index,
    sarima_predictions,
    label='Predicted'
)

plt.fill_between(
    df_test.index,
    conf_int.iloc[:, 0],
    conf_int.iloc[:, 1],
    alpha=0.2,
    label='95% Confidence Interval'
)

plt.xlabel('Month')
plt.ylabel('Thousands of Passengers')
plt.title('SARIMA: Actual vs Predicted')
plt.legend()
plt.show()
```

### B5. Evaluate

```python
# Compute RMSE
sarima_rmse = root_mean_squared_error(
    df_test['thousands_of_passengers'],
    sarima_predictions
)

print("SARIMA RMSE:", sarima_rmse)
```

### B6. ARIMA vs. SARIMA

Plot both sets of predictions against the actual test values side by side and print both RMSEs together.

```python
#YOUR CODE HERE
arima_rmse = root_mean_squared_error(
    df_test['thousands_of_passengers'],
    predictions
)

sarima_rmse = root_mean_squared_error(
    df_test['thousands_of_passengers'],
    sarima_predictions
)

print("ARIMA RMSE:", arima_rmse)
print("SARIMA RMSE:", sarima_rmse)
```

```python
plt.figure(figsize=(12, 6))

# Actual values
plt.plot(
    df_test.index,
    df_test['thousands_of_passengers'],
    label='Actual'
)

# ARIMA predictions
plt.plot(
    df_test.index,
    predictions,
    label='ARIMA'
)

# SARIMA predictions
plt.plot(
    df_test.index,
    sarima_predictions,
    label='SARIMA'
)

plt.xlabel('Month')
plt.ylabel('Thousands of Passengers')
plt.title('ARIMA vs SARIMA Predictions')
plt.legend()
plt.show()
```

## 2. Forecast the future

Retrain your **winning model** on the *full* dataset and forecast 12 months into the future

**Bonus**: *Plot the forecast with confidence intervals*.

```python
#YOUR CODE HERE
final_model = SARIMAX(
    df['thousands_of_passengers'],
    order=sarima_auto_model.order,
    seasonal_order=sarima_auto_model.seasonal_order
)

final_fit = final_model.fit()

# Forecast the next 12 months
future_forecast = final_fit.get_forecast(steps=12)

future_predictions = future_forecast.predicted_mean
future_conf_int = future_forecast.conf_int()

future_predictions
```

```python
plt.figure(figsize=(12, 6))

# Historical data
plt.plot(
    df.index,
    df['thousands_of_passengers'],
    label='Historical'
)

# Future forecast
plt.plot(
    future_predictions.index,
    future_predictions,
    label='12-Month Forecast'
)

# Confidence interval
plt.fill_between(
    future_predictions.index,
    future_conf_int.iloc[:, 0],
    future_conf_int.iloc[:, 1],
    alpha=0.2,
    label='95% Confidence Interval'
)

plt.xlabel('Month')
plt.ylabel('Thousands of Passengers')
plt.title('12-Month Future Passenger Forecast')
plt.legend()
plt.show()
```

## Takeaways
- Seasonality is easy to miss if you don't plot/decompose the data, hence, always look before you model.
- SARIMA isn't automatically the right call; it's justified when the seasonal component is real and the added parameters actually reduce error on held-out data.
- Always evaluate on a proper time-ordered train,test, and split.
